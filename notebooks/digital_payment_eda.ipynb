# Digital Payment Insight — EDA
This notebook accompanies Use Case 13. All records and review labels are synthetic; metrics are not real fraud-detection results.
Run `python src/build_project.py` from the repository root first to generate data and reports.

```python
from pathlib import Path
import json
import pandas as pd
ROOT = Path.cwd()
clean = pd.read_csv(ROOT / 'data/processed/cleaned_transactions.csv')
metrics = json.loads((ROOT / 'reports/model_metrics.json').read_text())
findings = json.loads((ROOT / 'reports/key_findings.json').read_text())
print(clean.shape)
display(clean.head())
print(json.dumps(metrics, indent=2))
print(json.dumps(findings, indent=2))
```

```python
import matplotlib.pyplot as plt
clean['amount'].plot(kind='hist', bins=50, title='Transaction amount distribution')
plt.xlabel('Transaction amount')
plt.show()
clean.groupby('hour').size().plot(kind='bar', title='Transaction volume by hour')
plt.xlabel('Hour')
plt.ylabel('Transactions')
plt.show()
print(clean.groupby('new_device_flag')['synthetic_review_label'].mean())
```
